```python
import numpy as np

A = np.arange(1, 21).reshape(4, 5)
print(A)
print(A.dtype)
print(A.ndim)
print(A.shape)
print(A.size)
print(A.itemsize)

B = A.reshape(5, 4)
print(B)
```

```python
import numpy as np

a = np.array([[0., 0., 0.],
              [10., 10., 10.],
              [20., 20., 20.],
              [30., 30., 30.]])

b = np.array([0., 1., 2.])

c = a + b
print(c)

mask = (c >= 22)
print(mask)

print(np.sum(c >= 22))
```

```python
import numpy as np

grades = np.array([[87, 96, 70],
                   [100, 87, 90],
                   [94, 77, 90],
                   [100, 81, 82]])

print(grades.sum())
print(grades.min())
print(grades.max())
print(grades.mean())
print(grades.std())
print(grades.var())

print(grades.mean(axis=0))
print(grades.mean(axis=1))
print(grades.max(axis=1))
```

```python
import numpy as np

arr = np.array([[2, 4, 6],
                [1, 3, 5]])

print(arr.sum(axis=1))
print(arr.sum(axis=0))
print(arr.mean())
```

```python
import numpy as np

A = np.array([[10, 20, 30],
              [40, 50, 60]])

B = np.array([1, 2, 3])

print(A + B)
print(A * 2)
print(A - 5)
```

```python
import numpy as np

grades = np.array([[70, 85, 90],
                   [88, 76, 95],
                   [60, 80, 72]])

print(np.unravel_index(np.argmax(grades), grades.shape))
print(np.argmax(grades, axis=1))
print(np.argmin(grades, axis=0))
```
